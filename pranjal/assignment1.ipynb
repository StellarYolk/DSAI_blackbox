### Task 1.1 — `parse_logs(logs)`
Return a dictionary grouping log messages by their level (`INFO`, `WARNING`, `ERROR`).

```python
def parse_logs(logs):
  grouped = {} # make an empty dictionary to hold logs #

  for log in logs: # go through each log one by one #

# here we will find closing bracket so we can slice out the log level like ERROR or INFO#


    end_idx = log.find("]")
    level = log[1:end_idx]
# finding thw dash and grab everything after it ss actual msg #
    dash_idx = log.find("-")
    msg = log[dash_idx + 3:]
# if we haven't the log level yet create a new list #
    if level in grouped:
      grouped[level] = []
# dropping the msg into the right category list #

      grouped [level].append(msg)

    return grouped
```

### Task 1.3 — `most_frequent_level(logs)`
Return the log level that appeared the most.

```python
def most_frequent_level(logs):
    counts = {} # to count how many times a level appears #

    for log in logs:
      end_idx = log.find("]")#slicing out the level#
      lvl = log[1:end_idx]

#if level is not in counts get we will increment it by 1 #
      if lvl not in counts:
        counts[lvl] = 1
      else:
        counts[lvl] += 1

      # finding which lvl has the highest count #
      max_lvl = none
      max_count = -1

      for lvl in counts:
        if counts[lvl] > max_count:
          max_lvl = lvl
          max_count = counts[lvl]

          return max_lvl
```

### Task 2.2 — Simulate the Task Scheduler
Insert all tasks into your `MinHeap` and process them in priority order.

```python
class MinHeap:
    def __init__(self):
        self.heap = []     # the heap lives in this list fr
        self.counter = 0   # counts insert order, used for tie breaking

    # helper: checks if a should come before b
    # a and b look like (priority, order, task)
    def _is_smaller(self, a, b):
        if a[0] != b[0]:
            return a[0] < b[0]   # lower priority number wins
        return a[1] < b[1]       # same priority? whoever came first goes first

    def insert(self, task):
        # task = (priority, name, duration)
        entry = (task[0], self.counter, task)
        self.counter += 1
        self.heap.append(entry)           # dump it at the end
        self._sift_up(len(self.heap) - 1) # then push it up to where it belongs

    def _sift_up(self, i):
        while i > 0:
            parent = (i - 1) // 2
            if self._is_smaller(self.heap[i], self.heap[parent]):
                # child is smaller than parent so we swap, no cap
                self.heap[i], self.heap[parent] = self.heap[parent], self.heap[i]
                i = parent
            else:
                break  # already in the right spot, we're done

    def extract_min(self):
        if self.is_empty():
            return None  # nothing to remove, heap is empty lol

        top = self.heap[0]        # the highest priority task
        last = self.heap.pop()    # take the last element out

        if len(self.heap) > 0:
            self.heap[0] = last   # put it at the root
            self._sift_down(0)    # let it sink to where it belongs

        return top[2]  # return the actual task tuple

    def _sift_down(self, i):
        n = len(self.heap)
        while True:
            smallest = i
            left = 2 * i + 1
            right = 2 * i + 2

            # check if left child is smaller
            if left < n and self._is_smaller(self.heap[left], self.heap[smallest]):
                smallest = left
            # check if right child is even smaller
            if right < n and self._is_smaller(self.heap[right], self.heap[smallest]):
                smallest = right

            if smallest == i:
                break  # nobody is smaller, heap property is chilling

            self.heap[i], self.heap[smallest] = self.heap[smallest], self.heap[i]
            i = smallest

    def peek(self):
        if self.is_empty():
            return None
        return self.heap[0][2]  # just look, don't remove

    def is_empty(self):
        return len(self.heap) == 0


# the tasks given in the question (priority, name, duration)
tasks = [
    (3, "Send weekly report", 15),
    (1, "Fix critical bug", 45),
    (2, "Code review", 30),
    (1, "Deploy hotfix", 20),
    (4, "Update documentation", 60),
]

# making the heap and throwing all the tasks in
pq = MinHeap()
for t in tasks:
    pq.insert(t)
```

## 🔷 Problem 3 — OOP: Library Management System

### Task 3.1 — Implement the `Book` class

```python
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_available = True   # every book starts out available ofc

    def __str__(self):
        # the nice readable version, this is what print() shows
        return f"'{self.title}' by {self.author} (ISBN: {self.isbn}) - Available: {self.is_available}"

    def __repr__(self):
        # for devs, shows how u would make this book again
        return f"Book('{self.title}', '{self.author}', '{self.isbn}')"
```

### Task 3.2 — Implement the `Member` class

```python
class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.borrowed_books = []   # starts empty, no books yet

    def borrow_book(self, book):
        # if someone already has it, we cant lend it out
        if not book.is_available:
            raise ValueError(f"'{book.title}' is already taken, sorry bestie")

        self.borrowed_books.append(book)
        book.is_available = False   # now its off the shelf

    def return_book(self, book):
        # cant return something u never borrowed lol
        if book not in self.borrowed_books:
            raise ValueError(f"{self.name} never borrowed '{book.title}'")

        self.borrowed_books.remove(book)
        book.is_available = True    # back on the shelf

    def __str__(self):
        return f"Member: {self.name} (ID: {self.member_id}) | Books borrowed: {len(self.borrowed_books)}"
```

### Task 3.3 — Implement the `Library` class

```python
class Library:
    def __init__(self, name):
        self.name = name
        self.books = []     # all the books the library owns
        self.members = []   # all the registered members

    def add_book(self, book):
        self.books.append(book)

    def register_member(self, member):
        self.members.append(member)

    def search_by_author(self, author):
        result = []
        for book in self.books:
            if book.author == author:
                result.append(book)
        return result

    def available_books(self):
        # only the books that are still on the shelf
        result = []
        for book in self.books:
            if book.is_available:
                result.append(book)
        return result

    def borrow_book(self, member_id, isbn):
        # first find the member and the book using their id / isbn
        member = None
        for m in self.members:
            if m.member_id == member_id:
                member = m

        book = None
        for b in self.books:
            if b.isbn == isbn:
                book = b

        # if either one doesnt exist then nothing to do here
        if member is None or book is None:
            print("member or book not found, check the id/isbn again")
            return

        # try/except so the program doesnt crash if the book is taken
        try:
            member.borrow_book(book)
            print(f"{member.name} borrowed '{book.title}'")
        except ValueError as e:
            print(f"Oops: {e}")

    def return_book(self, member_id, isbn):
        member = None
        for m in self.members:
            if m.member_id == member_id:
                member = m

        book = None
        for b in self.books:
            if b.isbn == isbn:
                book = b

        if member is None or book is None:
            print("member or book not found, check the id/isbn again")
            return

        try:
            member.return_book(book)
            print(f"{member.name} returned '{book.title}'")
        except ValueError as e:
            print(f"Oops: {e}")
```
