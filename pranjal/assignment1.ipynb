### Task 1.1 — `parse_logs(logs)`
Return a dictionary grouping log messages by their level (`INFO`, `WARNING`, `ERROR`).

```python
def parse_logs(logs):
  grouped = {} # make an empty dictionary to hold logs #

  for log in logs: # go through each log one by one #

# here we will find closing bracket so we can slice out the log level like ERROR or INFO#


    end_idx = log.find("]")
    level = log[1:end_idx]
# finding thw dash and grab everything after it ss actual msg #
    dash_idx = log.find("-")
    msg = log[dash_idx + 3:]
# if we haven't the log level yet create a new list #
    if level in grouped:
      grouped[level] = []
# dropping the msg into the right category list #

      grouped [level].append(msg)

    return grouped
```

### Task 1.3 — `most_frequent_level(logs)`
Return the log level that appeared the most.

```python
def most_frequent_level(logs):
    counts = {} # to count how many times a level appears #

    for log in logs:
      end_idx = log.find("]")#slicing out the level#
      lvl = log[1:end_idx]

#if level is not in counts get we will increment it by 1 #
      if lvl not in counts:
        counts[lvl] = 1
      else:
        counts[lvl] += 1

      # finding which lvl has the highest count #
      max_lvl = none
      max_count = -1

      for lvl in counts:
        if counts[lvl] > max_count:
          max_lvl = lvl
          max_count = counts[lvl]

          return max_lvl
```
