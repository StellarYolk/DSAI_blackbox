## 🔷 Problem 1 — Core Python: Log File Analyzer

### Given Data
Run this cell as-is. Do not modify the `logs` list.

```python
logs = [
    "[INFO] 2024-03-01 10:00:01 - Server started",
    "[ERROR] 2024-03-01 10:05:22 - Disk space low",
    "[WARNING] 2024-03-01 10:07:45 - CPU usage high",
    "[ERROR] 2024-03-01 10:12:10 - Failed to connect to database",
    "[INFO] 2024-03-01 10:15:00 - Backup completed",
    "[WARNING] 2024-03-02 09:00:00 - Memory usage above threshold",
    "[ERROR] 2024-03-02 09:45:11 - Timeout on request",
]
```

### Task 1.1 — `parse_logs(logs)`
Return a dictionary grouping log messages by their level (`INFO`, `WARNING`, `ERROR`).

```python
def parse_logs(logs):
    parsed_logs = {}
    for line in logs:
        level = line[1:line.index(']')] # Parsing log level
        message = line.split(" - ")[1] # Parsing log message
        if level not in parsed_logs:
            parsed_logs[level] = []
        parsed_logs[level].append(message)
    return parsed_logs

# Test
grouped = parse_logs(logs)
print("Grouped Logs:")
for level, messages in grouped.items():
    print(f"  {level}: {messages}")
```

### Task 1.2 — `error_summary(logs)`
Return the count of ERROR entries per date.

```python
def error_summary(logs):
    parsed_logs = {}
    for line in logs:
        date = line[(line.index(']') + 2):(line.index(']') + 12)] # Parsing date
        if date not in parsed_logs:
            parsed_logs[date] = 0
        level = line[1:line.index(']')] # determing log level
        if level == 'ERROR':
            parsed_logs[date] += 1 # Incrementing date counter
    return parsed_logs

# Test
print("Errors per date:", error_summary(logs))
```

### Task 1.3 — `most_frequent_level(logs)`
Return the log level that appeared the most.

```python
def most_frequent_level(logs):
    parsed_logs = parse_logs(logs) # sorting logs into log levels
    err_lvl = None
    err_freq = 0
    for level in parsed_logs:
        messages = parsed_logs[level]
        if len(messages) > err_freq: # determing max freq
            err_freq = len(messages)
            err_lvl = level
        elif len(messages) == err_freq: # determining same freq log level
            err_lvl += f', {level}'
    return err_lvl

# Test
print("Most frequent level:", most_frequent_level(logs))
```

## 🔷 Problem 2 — DSA: Task Scheduler using a Priority Queue

### Task 2.1 — Implement `MinHeap` class
Implement a Min-Heap based Priority Queue from scratch. **Do not import or use `heapq`.**

Each task is a tuple: `(priority, name, duration)`

```python
class MinHeap:
    def __init__(self):
        self.heap = []  # Internal list to store heap elements
        self.counter = 0 # A counter to determine FIFO order
    
    def insert(self, task):
        """
        Insert a task tuple (priority, name, duration) into the heap.
        Maintain the min-heap property after insertion.
        """
        priority, name, duration = task
        self.counter += 1 # incrementing the counter to determine FIFO order
        task = (priority, self.counter, task) # new task tuple with counter
        self.heap.append(task) # append the new insertion at last
        self._heapify_up(len(self.heap) - 1) # reordering the heap from the new task
    
    def extract_min(self):
        """
        Remove and return the task with the lowest priority number.
        Maintain the min-heap property after removal.
        """
        length = len(self.heap)
        if length != 0:
            priority, counter, max_priority_task = self.heap[0] # root will always be lowest priority number task tuple
            self.heap[0], self.heap[length - 1] = self.heap[length - 1], self.heap[0] # swapping with last element for clean exit
            self.heap.pop() # removing the max priority task
            self._heapify_down(0) # re-ordering the heap
            return max_priority_task
        else:
            return (None, None, None)
    
    def peek(self):
        """
        Return the task with the lowest priority number without removing it.
        """
        if len(self.heap) != 0:
            priority, counter, task = self.heap[0] # root will always be lowest priority number task tuple
            return task
        else:
            return (None, None, None)
    
    def is_empty(self):
        """
        Return True if the heap has no elements.
        """
        return (len(self.heap) == 0)
    
    def _heapify_up(self, index):
        """
        Helper: bubble the element at `index` up to restore heap property.

        Index is first checked if its the root, if True, the function exits.
        """
        if index > 0:
            parent = (index - 1) // 2
        else:
            return None
        while self.heap[parent] > self.heap[index] and index != 0: # Determining smaller priority number task tuple
            self.heap[parent], self.heap[index] = self.heap[index], self.heap[parent] # Swapping
            index = parent # new target index
            parent = (index - 1) // 2 # new parent
    
    def _heapify_down(self, index):
        """
        Helper: push the element at `index` down to restore heap property.
        """
        length = len(self.heap)
        left = ((2 * index) + 1) if ((2 * index) + 1) < length else None # left child of parent (index)
        right = ((2 * index) + 2) if ((2 * index) + 2) < length else None # right child of parent (index)
        while True:
            small_index = index # the index of smaller priority number task tuple
            if (left is not None and right is not None) and (self.heap[index] > self.heap[left]) and (self.heap[right] > self.heap[left]): # Determining smaller child
                small_index = left
            elif (left is not None and right is not None) and (self.heap[index] > self.heap[right]) and (self.heap[left] > self.heap[right]): # Determining smaller child
                small_index = right
            elif (left is None and right is not None) and (self.heap[index] > self.heap[right]):
                small_index = right
            elif (left is not None and right is None) and (self.heap[index] > self.heap[left]):
                small_index = left
            else:
                break
            self.heap[small_index], self.heap[index] = self.heap[index], self.heap[small_index] # swapping parent with the smaller child
            index = small_index # updating the new index to be pushed down
            left = ((2 * index) + 1) if ((2 * index) + 1) < length else None # new left child
            right = ((2 * index) + 2) if ((2 * index) + 2) < length else None # new right child
```

## 🔷 Problem 3 — OOP: Library Management System

### Task 3.1 — Implement the `Book` class

```python
class Book:
    def __init__(self, title, author, isbn):
        # Attributes: title, author, isbn, is_available (default True)
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_available = True
    
    def __str__(self):
        return f"'{self.title}' by {self.author} (ISBN: {self.isbn}) - Available: {self.is_available}"
    
    def __repr__(self):
        return f"Title={self.title};author={self.author};isbn={self.isbn};Available={self.is_available}"
```

### Task 3.2 — Implement the `Member` class

```python
class Member:
    def __init__(self, name, member_id):
        # Attributes: name, member_id, borrowed_books (empty list)
        self.name = name
        self.member_id = member_id
        self.borrowed_books = []

    def borrow_book(self, book):
        """
        Borrow a book if it's available.
        Raise an exception (e.g., ValueError) if the book is not available.
        Update book.is_available to False.
        """
        if book.is_available:
            self.borrowed_books.append(book)
            book.is_available = False
        else:
            raise ValueError("Book not available")
    
    def return_book(self, book):
        """
        Return a borrowed book.
        Remove it from borrowed_books and set book.is_available to True.
        """
        self.borrowed_books.remove(book)
        book.is_available = True
    
    def __str__(self):
        return f"Member: {self.name} (ID: {self.member_id}) | Books borrowed: {len(self.borrowed_books)}"
```

### Task 3.3 — Implement the `Library` class

```python
class Library:
    def __init__(self, name):
        # Attributes: name, books (empty list), members (empty list)
        self.name = name
        self.books = []
        self.members = []
    
    def add_book(self, book):
        """Add a Book object to the library's collection."""
        self.books.append(book)
    
    def register_member(self, member):
        """Register a Member to the library."""
        self.members.append(member)
    
    def search_by_author(self, author):
        """Return a list of all books by the given author."""
        author_books = []
        for book in self.books:
            if author in book.author:
                author_books.append(book)
        return author_books
    
    def available_books(self):
        """Return a list of all currently available books."""
        available_books = []
        for book in self.books:
            if book.is_available == True:
                available_books.append(book)
        return available_books
    
    def borrow_book(self, member_id, isbn):
        """
        Find the member and book by ID/ISBN.
        Call member.borrow_book(book) to handle the transaction.
        Handle any exceptions gracefully.
        """
        try:
            borrower = None
            target_book = None
            for member in self.members:
                if member.member_id == member_id:
                    borrower = member
                    break
            for book in self.books:
                if book.isbn == isbn:
                    target_book = book
                    break
            borrower.borrow_book(target_book)
        except Exception as e:
            return e

    def return_book(self, member_id, isbn):
        """
        Find the member and book by ID/ISBN.
        Call member.return_book(book) to handle the transaction.
        """
        try:
            borrower = None
            target_book = None
            for member in self.members:
                if member.member_id == member_id:
                    borrower = member
                    break
            for book in self.books:
                if book.isbn == isbn:
                    target_book = book
                    break
            borrower.return_book(target_book)
        except Exception as e:
            return e
```
