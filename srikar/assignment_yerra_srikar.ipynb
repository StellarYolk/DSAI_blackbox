## 🔷 Problem 1 — Core Python: Log File Analyzer

### Given Data
Run this cell as-is. Do not modify the `logs` list.

```python
logs = [
    "[INFO] 2024-03-01 10:00:01 - Server started",
    "[ERROR] 2024-03-01 10:05:22 - Disk space low",
    "[WARNING] 2024-03-01 10:07:45 - CPU usage high",
    "[ERROR] 2024-03-01 10:12:10 - Failed to connect to database",
    "[INFO] 2024-03-01 10:15:00 - Backup completed",
    "[WARNING] 2024-03-02 09:00:00 - Memory usage above threshold",
    "[ERROR] 2024-03-02 09:45:11 - Timeout on request",
]
```

### Task 1.1 — `parse_logs(logs)`
Return a dictionary grouping log messages by their level (`INFO`, `WARNING`, `ERROR`).

```python
def parse_logs(logs):
    # Creating an empty dictionery
    grouped_logs = {}

    # Loop through each log entry in the list
    for log in logs:
      # Extract the log level (text between '[' and ']')
        start = log.find("[")
        end = log.find("]")
        level = log[start + 1 : end]

        # Extract the message (text after ' - ')
        dash_index = log.find(" - ")  # it returns the index of the space before the dash.
        message = log[dash_index + 3 :]

        # Check if the log level exist in dictionary
        if level not in grouped_logs:
            # If not, create a new list for this level
            grouped_logs[level] = []  # to store multiple messages

        # Add the message to the list for respective levels
        grouped_logs[level].append(message)

    # Return the dictionary after processing all logs
    return grouped_logs

# Test
grouped = parse_logs(logs)
print("Grouped Logs:")
for level, messages in grouped.items():
    print(f"  {level}: {messages}")
```

### Task 1.2 — `error_summary(logs)`
Return the count of ERROR entries per date.

```python
def error_summary(logs):
    # Create an empty dictionary 
    error_counts = {}

    # Loop through each log entry in the list
    for log in logs:
        # Extract the log level to check if it's an ERROR
        start = log.find("[")
        end = log.find("]")
        level = log[start + 1 : end]

        # Only process entries that are error's
        if level == "ERROR":
            # Extract the date (YYYY-MM-DD starts right after '] ' which is end + 2)
            date_start = end + 2
            date = log[date_start : date_start + 10]

            # Count the occurrences of each date
            if date not in error_counts:
                # First time seeing this error date, start count at 1
                error_counts[date] = 1
            else:
                # Date already exists, increment its count by 1
                error_counts[date] += 1

    # Return the dictionary of error counts per date
    return error_counts

# Test
print("Errors per date:", error_summary(logs))
```

### Task 1.3 — `most_frequent_level(logs)`
Return the log level that appeared the most.

```python
def most_frequent_level(logs):
    # Create an empty dictionary
    level_counts = {}

    # Loop through each log entry to extract and count levels
    for log in logs:
        start_bracket = log.find("[")
        end_bracket = log.find("]")
        level = log[start_bracket + 1 : end_bracket]

        # Add or update the count 
        if level not in level_counts:
            level_counts[level] = 1
        else:
            level_counts[level] += 1

    # Find the level with the highest count
    most_frequent = None
    max_count = 0

    for level, count in level_counts.items():
        if count > max_count:
            max_count = count
            most_frequent = level

    # Step 4: Return the log level that appeared the most
    return most_frequent




# Test
print("Most frequent level:", most_frequent_level(logs))
```

## 🔷 Problem 2 — DSA: Task Scheduler using a Priority Queue

### Task 2.1 — Implement `MinHeap` class
Implement a Min-Heap based Priority Queue from scratch. **Do not import or use `heapq`.**

Each task is a tuple: `(priority, name, duration)`

```python
class MinHeap:

    def __init__(self):
        self.heap = []  # Internal list to store heap elements

    def insert(self, task):
        """Insert a task tuple (priority, name, duration) into the heap.

        Maintain the min-heap property after insertion.
        """
        # Append the new task to the end of the heap list
        self.heap.append(task)
        # Bubble the newly added element up to its correct position
        self._heapify_up(len(self.heap) - 1)

    def extract_min(self):
        """Remove and return the task with the lowest priority number.

        Maintain the min-heap property after removal.
        """
        if self.is_empty():
            return None

        # If there is only one element, remove and return it
        if len(self.heap) == 1:
            return self.heap.pop()

        # Save the minimum element (root)
        min_task = self.heap[0]

        # Move the last element to the root position
        self.heap[0] = self.heap.pop()

        # Push the new root down to restore the min-heap property
        self._heapify_down(0)

        return min_task

    def peek(self):
        """Return the task with the lowest priority number without removing
        it."""
        if self.is_empty():
            return None
        return self.heap[0]

    def is_empty(self):
        """Return True if the heap has no elements."""
        return len(self.heap) == 0

    def _heapify_up(self, index):
        """Helper: bubble the element at `index` up to restore heap property."""
        while index > 0:
            parent_index = (index - 1) // 2

            # If current element is smaller than its parent, swap them
            if self.heap[index] < self.heap[parent_index]:
                self.heap[index], self.heap[parent_index] = (
                    self.heap[parent_index],
                    self.heap[index],
                )
                index = parent_index
            else:
                break

    def _heapify_down(self, index):
        """Helper: push the element at `index` down to restore heap property."""
        size = len(self.heap)

        while True:
            smallest = index
            left_child = 2 * index + 1
            right_child = 2 * index + 2

            # Check if left child exists and is smaller than current smallest
            if (
                left_child < size
                and self.heap[left_child] < self.heap[smallest]
            ):
                smallest = left_child

            # Check if right child exists and is smaller than current smallest
            if (
                right_child < size
                and self.heap[right_child] < self.heap[smallest]
            ):
                smallest = right_child

            # If the smallest element is not the current node, swap and continue
            if smallest != index:
                self.heap[index], self.heap[smallest] = (
                    self.heap[smallest],
                    self.heap[index],
                )
                index = smallest
            else:
                break
```

## 🔷 Problem 3 — OOP: Library Management System

### Task 3.1 — Implement the `Book` class

```python
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        # Set availability to True by default
        self.is_available = True
    
    def __str__(self):
        return f"'{self.title}' by {self.author} (ISBN: {self.isbn}) - Available: {self.is_available}"
    
    def __repr__(self):
        return f"Book('{self.title}', '{self.author}', '{self.isbn}')"
        
```

### Task 3.2 — Implement the `Member` class

```python
class Member:

    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.borrowed_books = []

    def borrow_book(self, book):
        """Borrow a book if it's available."""
        # Check if the book is available
        if not book.is_available:
            raise ValueError(f"The book '{book.title}' is currently unavailable.")

        # Mark book as unavailable
        book.is_available = False

        # Add book to member's borrowed list
        self.borrowed_books.append(book)

    def return_book(self, book):
        """Return a borrowed book."""
        # Check if the member actually borrowed this book
        if book in self.borrowed_books:
            # Remove book from borrowed list
            self.borrowed_books.remove(book)
            # Mark book as available again
            book.is_available = True

    def __str__(self):
        borrowed_count = len(self.borrowed_books)
        return f"Member: {self.name} (ID: {self.member_id}) | Books borrowed: {borrowed_count}"
```

### Task 3.3 — Implement the `Library` class

```python
class Library:

    def __init__(self, name):
        # Store library name and set up empty lists for books and members
        self.name = name
        self.books = []
        self.members = []

    def add_book(self, book):
        """Add a Book object to the library's collection."""
        self.books.append(book)

    def register_member(self, member):
        """Register a Member to the library."""
        self.members.append(member)

    def search_by_author(self, author):
        """Return a list of all books by the given author."""
        result = []
        for book in self.books:
            if book.author.lower() == author.lower():
                result.append(book)
        return result

    def available_books(self):
        """Return a list of all currently available books."""
        available_list = []
        for book in self.books:
            if book.is_available:
                available_list.append(book)
        return available_list

    def borrow_book(self, member_id, isbn):
        """Find the member and book by ID/ISBN and handle borrowing."""
        # Find the member by ID
        target_member = None
        for member in self.members:
            if member.member_id == member_id:
                target_member = member
                break

        # Find the book by ISBN
        target_book = None
        for book in self.books:
            if book.isbn == isbn:
                target_book = book
                break

        # If both member and book exist, try to borrow
        if target_member and target_book:
            try:
                target_member.borrow_book(target_book)
                print(
                    f"Success: '{target_book.title}' borrowed by {target_member.name}."
                )
            except ValueError as error:
                # Handle unavailable book exception gracefully
                print(f"Error: {error}")
        else:
            print("Error: Member or Book not found.")

    def return_book(self, member_id, isbn):
        """Find the member and book by ID/ISBN and handle returning."""
        # Find the member by ID
        target_member = None
        for member in self.members:
            if member.member_id == member_id:
                target_member = member
                break

        # Find the book by ISBN
        target_book = None
        for book in self.books:
            if book.isbn == isbn:
                target_book = book
                break

        # If both member and book exist, process the return
        if target_member and target_book:
            target_member.return_book(target_book)
            print(
                f"Success: '{target_book.title}' returned by {target_member.name}."
            )
        else:
            print("Error: Member or Book not found.")
```
